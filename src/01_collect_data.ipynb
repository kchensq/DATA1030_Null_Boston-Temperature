```python
import json

import pandas as pd
import requests

URL = "https://archive-api.open-meteo.com/v1/archive"
PARAMS = {
    "latitude": 42.361145,
    "longitude": -71.057083,
    "start_date": "2006-01-01",
    "end_date": "2025-12-31",
    "daily": ["temperature_2m_mean", "precipitation_sum", "wind_speed_10m_max", "snowfall_sum"],
    "models": "era5",
    "timezone": "America/New_York",
    "temperature_unit": "celsius",
    "wind_speed_unit": "kmh",
    "precipitation_unit": "mm",
}

response = requests.get(URL, params=PARAMS, timeout=120)
response.raise_for_status()
data = response.json()
if data.get("error"):
    raise RuntimeError(data.get("reason"))

with open("../data/boston_weather_raw.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)
df = pd.DataFrame(data["daily"])
df.to_csv("../data/boston_weather_raw.csv", index=False)

print(f"Downloaded {len(df):,} daily records, {df['time'].min()} to {df['time'].max()}")
```
